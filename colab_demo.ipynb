## ⚙️ Machine Learning Pipeline Demo

Let's test the machine learning capabilities with sample data and model training.

```python
# Machine Learning Pipeline Demo
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification, make_regression
import matplotlib.pyplot as plt
import seaborn as sns

print("🤖 Machine Learning Pipeline Demo")
print("=" * 50)

# Generate sample classification dataset
print("\n📊 Generating Sample Datasets:")
X_class, y_class = make_classification(
    n_samples=1000, 
    n_features=20, 
    n_classes=3, 
    n_informative=15,
    random_state=42
)

X_reg, y_reg = make_regression(
    n_samples=1000, 
    n_features=15, 
    n_targets=1, 
    random_state=42
)

print(f"  Classification Data: {X_class.shape}, {y_class.shape}")
print(f"  Regression Data: {X_reg.shape}, {y_reg.shape}")

# Train classification models
print("\n🎯 Training Classification Models:")
class_results = orchestrator.train_sklearn_models(
    X_class, y_class, 
    ['random_forest', 'svm', 'logistic_regression']
)

print("Classification Results:")
for model_name, result in class_results.items():
    if 'error' not in result:
        print(f"  ✅ {model_name}: Trained successfully")
    else:
        print(f"  ❌ {model_name}: {result['error']}")

# Train regression models
print("\n📈 Training Regression Models:")
reg_results = orchestrator.train_sklearn_models(
    X_reg, y_reg,
    ['random_forest', 'linear_regression', 'ridge']
)

print("Regression Results:")
for model_name, result in reg_results.items():
    if 'error' not in result:
        print(f"  ✅ {model_name}: Trained successfully")
    else:
        print(f"  ❌ {model_name}: {result['error']}")

# Create visualization
print("\n📊 Creating Model Performance Visualization:")
plt.figure(figsize=(12, 8))

# Plot sample data
plt.subplot(2, 2, 1)
plt.scatter(X_class[:, 0], X_class[:, 1], c=y_class, cmap='viridis', alpha=0.6)
plt.title('Classification Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

plt.subplot(2, 2, 2)
plt.scatter(X_reg[:, 0], y_reg, alpha=0.6)
plt.title('Regression Data')
plt.xlabel('Feature 1')
plt.ylabel('Target')

plt.subplot(2, 2, 3)
model_names = list(class_results.keys())
success_count = sum(1 for r in class_results.values() if 'error' not in r)
plt.bar(model_names, [1]*len(model_names), color=['green' if 'error' not in class_results[m] else 'red' for m in model_names])
plt.title('Classification Models')
plt.ylabel('Status')
plt.xticks(rotation=45)

plt.subplot(2, 2, 4)
reg_model_names = list(reg_results.keys())
reg_success_count = sum(1 for r in reg_results.values() if 'error' not in r)
plt.bar(reg_model_names, [1]*len(reg_model_names), color=['green' if 'error' not in reg_results[m] else 'red' for m in reg_model_names])
plt.title('Regression Models')
plt.ylabel('Status')
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig('ml_pipeline_results.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"\n✅ ML Pipeline completed! {success_count + reg_success_count} models trained successfully.")
```

## 🧠 Deep Learning Models Demo

Now let's create and test deep learning models using TensorFlow/Keras.

```python
# Deep Learning Models Demo
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt

print("🧠 Deep Learning Models Demo")
print("=" * 50)

# Create CNN for image classification
print("\n🖼️ Creating CNN Image Classifier:")
try:
    cnn_model = orchestrator.create_deep_learning_model(
        'image_classifier',
        input_shape=(64, 64, 3),
        num_classes=10,
        model_name='demo_cnn'
    )
    print("✅ CNN created successfully")
    print(f"  Input shape: {cnn_model.input_shape}")
    print(f"  Output shape: {cnn_model.output_shape}")
    print(f"  Total parameters: {cnn_model.count_params():,}")
except Exception as e:
    print(f"❌ CNN creation failed: {str(e)}")

# Create LSTM for text processing
print("\n📝 Creating LSTM Text Classifier:")
try:
    lstm_model = orchestrator.create_deep_learning_model(
        'text_classifier',
        vocab_size=1000,
        embedding_dim=100,
        max_length=50,
        num_classes=5,
        model_name='demo_lstm'
    )
    print("✅ LSTM created successfully")
    print(f"  Input shape: {lstm_model.input_shape}")
    print(f"  Output shape: {lstm_model.output_shape}")
    print(f"  Total parameters: {lstm_model.count_params():,}")
except Exception as e:
    print(f"❌ LSTM creation failed: {str(e)}")

# Create GAN components
print("\n🎨 Creating GAN Components:")
try:
    generator = orchestrator.create_deep_learning_model(
        'gan_generator',
        latent_dim=100,
        model_name='demo_generator'
    )
    
    discriminator = orchestrator.create_deep_learning_model(
        'gan_discriminator',
        model_name='demo_discriminator'
    )
    
    print("✅ GAN components created successfully")
    print(f"  Generator parameters: {generator.count_params():,}")
    print(f"  Discriminator parameters: {discriminator.count_params():,}")
except Exception as e:
    print(f"❌ GAN creation failed: {str(e)}")

# Visualize model architectures
print("\n📊 Model Architecture Summary:")
if 'cnn_model' in locals():
    print("\n🖼️ CNN Architecture:")
    cnn_model.summary()

if 'lstm_model' in locals():
    print("\n📝 LSTM Architecture:")
    lstm_model.summary()

print("\n✅ Deep Learning Models Demo completed!")
```

## 📊 Data Analysis & Visualization Demo

Let's test the data analysis capabilities with automated insights and visualizations.

```python
# Data Analysis & Visualization Demo
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("📊 Data Analysis & Visualization Demo")
print("=" * 50)

# Generate sample business data
print("\n📈 Generating Sample Business Data:")
np.random.seed(42)
dates = pd.date_range('2024-01-01', periods=365, freq='D')

sample_data = pd.DataFrame({
    'date': dates,
    'sales': np.random.normal(5000, 1000, 365).cumsum(),
    'customers': np.random.normal(100, 20, 365).cumsum(),
    'marketing_spend': np.random.normal(500, 100, 365),
    'website_visits': np.random.normal(10000, 2000, 365),
    'conversion_rate': np.random.beta(2, 5, 365) * 0.1
})

# Add some trends
sample_data['sales'] += np.arange(365) * 10
sample_data['customers'] += np.arange(365) * 0.5

print(f"  Dataset shape: {sample_data.shape}")
print(f"  Date range: {sample_data['date'].min()} to {sample_data['date'].max()}")
print(f"  Columns: {list(sample_data.columns)}")

# Analyze data using orchestrator
print("\n🔍 Analyzing Data with AI Orchestrator:")

# Save data for analysis
sample_data.to_csv('data/sample_business_data.csv', index=False)

# Perform data analysis
analysis_result = orchestrator.analyze_data('data/sample_business_data.csv')

if 'error' not in analysis_result:
    print("✅ Data analysis completed successfully")
    print(f"  Summary: {analysis_result.get('summary', 'N/A')[:200]}...")
else:
    print(f"❌ Data analysis failed: {analysis_result.get('error')}")

# Create comprehensive visualizations
print("\n📊 Creating Data Visualizations:")

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('Business Data Analysis Dashboard', fontsize=16, fontweight='bold')

# Sales trend
axes[0, 0].plot(sample_data['date'], sample_data['sales'], color='blue', linewidth=2)
axes[0, 0].set_title('Sales Trend')
axes[0, 0].set_xlabel('Date')
axes[0, 0].set_ylabel('Sales ($)')
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].tick_params(axis='x', rotation=45)

# Customer growth
axes[0, 1].plot(sample_data['date'], sample_data['customers'], color='green', linewidth=2)
axes[0, 1].set_title('Customer Growth')
axes[0, 1].set_xlabel('Date')
axes[0, 1].set_ylabel('Customers')
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].tick_params(axis='x', rotation=45)

# Marketing spend distribution
axes[0, 2].hist(sample_data['marketing_spend'], bins=30, color='orange', alpha=0.7)
axes[0, 2].set_title('Marketing Spend Distribution')
axes[0, 2].set_xlabel('Marketing Spend ($)')
axes[0, 2].set_ylabel('Frequency')
axes[0, 2].grid(True, alpha=0.3)

# Website visits trend
axes[1, 0].plot(sample_data['date'], sample_data['website_visits'], color='purple', linewidth=2)
axes[1, 0].set_title('Website Visits Trend')
axes[1, 0].set_xlabel('Date')
axes[1, 0].set_ylabel('Visits')
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].tick_params(axis='x', rotation=45)

# Conversion rate trend
axes[1, 1].plot(sample_data['date'], sample_data['conversion_rate'], color='red', linewidth=2)
axes[1, 1].set_title('Conversion Rate Trend')
axes[1, 1].set_xlabel('Date')
axes[1, 1].set_ylabel('Conversion Rate')
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].tick_params(axis='x', rotation=45)

# Correlation heatmap
numeric_cols = ['sales', 'customers', 'marketing_spend', 'website_visits', 'conversion_rate']
correlation_matrix = sample_data[numeric_cols].corr()
im = axes[1, 2].imshow(correlation_matrix, cmap='coolwarm', aspect='auto')
axes[1, 2].set_title('Correlation Matrix')
axes[1, 2].set_xticks(range(len(numeric_cols)))
axes[1, 2].set_yticks(range(len(numeric_cols)))
axes[1, 2].set_xticklabels(numeric_cols, rotation=45, ha='right')
axes[1, 2].set_yticklabels(numeric_cols)

# Add colorbar for correlation
cbar = plt.colorbar(im, ax=axes[1, 2])
cbar.set_label('Correlation Coefficient')

plt.tight_layout()
plt.savefig('data_analysis_dashboard.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n✅ Data analysis and visualization completed!")
print("📊 Dashboard saved as: data_analysis_dashboard.png")
```

## 🔄 Integrated AI Workflow Demo

Now let's test the complete integrated workflow that combines multiple AI components for a comprehensive solution.

```python
# Integrated AI Workflow Demo
print("🔄 Integrated AI Workflow Demo")
print("=" * 50)

# Test different types of integrated requests
test_requests = [
    "I need help with my account login issue",
    "Generate a monthly sales report",
    "Summarize the latest customer feedback",
    "Analyze quarterly performance data",
    "Create predictions for customer churn"
]

print("🚀 Processing Integrated AI Requests:\n")

for i, request in enumerate(test_requests, 1):
    print(f"{i}. 📝 Request: {request}")
    
    # Process through integrated workflow
    result = orchestrator.run_complete_ai_workflow(request)
    
    if 'error' not in result:
        print(f"   ✅ Workflow: {result.get('workflow_type', 'unknown')}")
        print(f"   🤖 JARVIS Response: {result.get('jarvis_response', 'N/A')[:100]}...")
        
        if 'status' in result:
            print(f"   📊 Status: {result['status']}")
        
        if 'response' in result:
            print(f"   💬 Response: {result['response'][:100]}...")
    else:
        print(f"   ❌ Error: {result.get('error', 'Unknown error')}")
    
    print()

print("✅ Integrated AI Workflow Demo completed!")
print("\n🎯 Key Features Demonstrated:")
  • Automatic request categorization
  • Multi-component coordination
  • Context-aware responses
  • Workflow automation
  • Intelligent routing to appropriate AI components
```

## 🌐 API Testing Demo

Let's test the REST API endpoints to demonstrate how the integrated system works via HTTP requests.

```python
# API Testing Demo
print("🌐 API Testing Demo")
print("=" * 50)

# Since we can't run the full API server in Colab, we'll demonstrate
# the API interface and show example requests

import json
import requests
from datetime import datetime

# Example API endpoints and payloads
api_examples = {
    'email_processing': {
        'url': 'POST /email/process',
        'payload': {
            'content': 'I need help with my account access',
            'sender': 'customer@example.com',
            'subject': 'Account Issue'
        },
        'description': 'Process email and generate auto-response'
    },
    
    'report_generation': {
        'url': 'POST /report/generate',
        'payload': {
            'data': {'sales': [1000, 1500, 2000], 'customers': [50, 75, 100]},
            'report_type': 'monthly',
            'title': 'Sales Performance Report'
        },
        'description': 'Generate automated business reports'
    },
    
    'text_summarization': {
        'url': 'POST /summarize/text',
        'payload': {
            'text': 'Long business document text...',
            'summary_type': 'executive'
        },
        'description': 'Summarize long documents'
    },
    
    'customer_support': {
        'url': 'POST /support/ticket',
        'payload': {
            'customer_email': 'user@company.com',
            'subject': 'Critical System Issue',
            'message': 'System is down, need immediate help',
            'priority': 'critical'
        },
        'description': 'Process customer support tickets'
    },
    
    'jarvis_command': {
        'url': 'POST /jarvis/command',
        'payload': {
            'command': 'Analyze customer data and create predictions'
        },
        'description': 'Process voice commands through JARVIS'
    },
    
    'ml_training': {
        'url': 'POST /ml/train',
        'payload': {
            'model_type': 'classification',
            'data': {'X': [[1,2,3], [4,5,6]], 'y': [0, 1]},
            'algorithms': ['random_forest', 'svm']
        },
        'description': 'Train machine learning models'
    }
}

print("📋 Available API Endpoints:\n")

for endpoint_name, details in api_examples.items():
    print(f"🔗 {endpoint_name.replace('_', ' ').title()}")
    print(f"   URL: {details['url']}")
    print(f"   Description: {details['description']}")
    print(f"   Example Payload: {json.dumps(details['payload'], indent=6)[:200]}...")
    print()

# Example curl commands
print("📝 Example API Usage (curl commands):\n")

curl_examples = [
    '# Email Processing',
    'curl -X POST http://localhost:8000/email/process \\',
    '  -H "Content-Type: application/json" \\',
    '  -d \'{"content": "Help with account", "sender": "user@example.com"}\'',
    '',
    '# Report Generation',
    'curl -X POST http://localhost:8000/report/generate \\',
    '  -H "Content-Type: application/json" \\',
    '  -d \'{"data": {"sales": [1000]}, "report_type": "monthly"}\'',
    '',
    '# JARVIS Command',
    'curl -X POST http://localhost:8000/jarvis/command \\',
    '  -H "Content-Type: application/json" \\',
    '  -d \'{"command": "Analyze sales data"}\''
]

for example in curl_examples:
    print(example)

print("\n✅ API Testing Demo completed!")
print("\n🚀 To run the actual API server:")
print("   python main.py --api")
print("   Then visit: http://localhost:8000/docs")
```
